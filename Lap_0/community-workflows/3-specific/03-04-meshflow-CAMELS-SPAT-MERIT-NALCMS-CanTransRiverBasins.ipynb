Sort the geofabric based on the drainage database rank prepared in advance, in order to reduce computation during the model-agnostic step. This pre-sorting also eliminates the need to re-sort the remapped forcing data from Easymore during the model-specific process.

```python
import os # built-in Python 3.10.2
import geopandas as gpd # version 0.14.3
import xarray as xr
import pandas as pd

# === Configuration ===
output_path = '/home/zelalem/github-repos/community-workflows/1-geofabric/CAMELS-SPAT/merged_lamped'
agg_basin = gpd.read_file(os.path.join(output_path, 'agg_CAMELS-MERIT_CanTrans_subbasins.shp'))
agg_river = gpd.read_file(os.path.join(output_path, 'agg_CAMELS-MERIT_CanTrans_rivers.shp'))
rank_column = "Rank"                       # Name of the column to sort by
ddb = xr.open_dataset(os.path.join('/home/zelalem/github-repos/camels_mesh_outputs', 'MESH_drainage_database.nc'))[['subbasin', 'Rank', 'Next']]

# This flattens the data into tabular form
df_netcdf = ddb.to_dataframe().reset_index()

# --- Merge on subbasin/COMID ---
agg_basin = agg_basin.merge(df_netcdf, left_on='COMID', right_on='subbasin')
agg_river = agg_river.merge(df_netcdf, left_on='COMID', right_on='subbasin')

# === Sort by the Rank column and drop extra column===
agg_basin_sorted = agg_basin.sort_values(by=rank_column, ascending=True)
agg_river_sorted = agg_river.sort_values(by=rank_column, ascending=True)
agg_basin_sorted = agg_basin_sorted.drop(columns=['subbasin'])
agg_river_sorted = agg_river_sorted.drop(columns=['subbasin'])

# if integer doesnot work convert them into string
#agg_basin_sorted["COMID"] = agg_basin_sorted["COMID"].astype(str)
#agg_river_sorted["COMID"] = agg_river_sorted["COMID"].astype(str)
#agg_river_sorted["NextDownID"] = agg_river_sorted["NextDownID"].astype(str)

# === Save to new shapefile ===
agg_basin_sorted.to_file(os.path.join(output_path, 'sorted_agg_CAMELS-MERIT_CanTrans_subbasins.shp'))
agg_river_sorted.to_file(os.path.join(output_path, 'sorted_agg_CAMELS-MERIT_CanTrans_rivers.shp'))
```
