```python
import netCDF4
import numpy as np
import matplotlib.pyplot as plt

def plot_netcdf(file_path, variable_name):
    # Open the NetCDF file
    dataset = netCDF4.Dataset(file_path, 'r')
    
    # Print out dataset information
    print(f"Dataset: {file_path}")
    print("Dimensions:")
    for dim in dataset.dimensions.values():
        print(f"  {dim.name}: {dim.size}")
    
    print("\nVariables:")
    for var in dataset.variables.values():
        print(f"  {var.name}: {var.dimensions}, {var.shape}")
    
    # Check if the variable exists in the dataset
    if variable_name not in dataset.variables:
        print(f"Variable '{variable_name}' not found in the dataset.")
        return
    
    # Read the variable data
    variable_data = dataset.variables[variable_name][:]
    
    # Close the dataset
    dataset.close()
    
    # Plotting
    plt.figure(figsize=(10, 6))
    
    # Handle different dimensions
    if len(variable_data.shape) == 1:
        plt.plot(variable_data)
        plt.xlabel('Index')
        plt.ylabel(variable_name)
        plt.title(f'Plot of {variable_name}')
    elif len(variable_data.shape) == 2:
        plt.imshow(variable_data, cmap='viridis', aspect='auto')
        plt.colorbar(label=variable_name)
        plt.title(f'Heatmap of {variable_name}')
    else:
        print("Unsupported variable shape for plotting.")
        return
    
    plt.show()

# Example usage
file_path = 'path_to_your_netcdf_file.nc'
variable_name = 'your_variable_name'
plot_netcdf(file_path, variable_name)
```
